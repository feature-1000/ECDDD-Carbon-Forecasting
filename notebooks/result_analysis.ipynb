# Result Analysis

Run the experiment scripts first, then use this notebook to inspect generated CSV files.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path('..').resolve()
RESULTS = ROOT / 'outputs' / 'results'
FIGURES = ROOT / 'outputs' / 'figures'
RESULTS, FIGURES
```

```python
forecast = pd.read_csv(RESULTS / 'forecast_metrics.csv')
forecast.sort_values(['Dataset', 'RMSE']).head(20)
```

```python
partition = pd.read_csv(RESULTS / 'dataset_partition_summary.csv')
partition
```

```python
ablation_path = RESULTS / 'ablation_results.csv'
if ablation_path.exists():
    ablation = pd.read_csv(ablation_path)
    display(ablation.sort_values('RMSE'))
```

```python
complexity_path = RESULTS / 'complexity_results.csv'
if complexity_path.exists():
    complexity = pd.read_csv(complexity_path)
    display(complexity)
```

```python
pred_path = RESULTS / 'forecast_predictions.csv'
if pred_path.exists():
    pred = pd.read_csv(pred_path)
    first_dataset = pred['Dataset'].iloc[0]
    sub = pred[pred['Dataset'] == first_dataset].drop(columns=['Dataset'])
    sub.plot(figsize=(12, 4), linewidth=1)
    plt.title(first_dataset)
    plt.tight_layout()
```
